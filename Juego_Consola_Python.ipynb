```python
import random

# Jugador
class Jugador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.salud = 100
        self.nivel = 1
        self.experiencia = 0

    def atacar(self):
        return random.randint(10, 20) * self.nivel

    def recibir_dano(self, dano):
        self.salud -= dano
        if self.salud <= 0:
            self.salud = 0
            print(f"{self.nombre} ha muerto")

    def ganar_experiencia(self, experiencia):
        print(f"{self.nombre} ha ganado {experiencia} puntos de experiencia")
        self.experiencia += experiencia
        if self.experiencia >= 100:
            self.nivel += 1
            self.experiencia = 0
            print(f"{self.nombre} ha subido de nivel a {self.nivel}")
```

```python
# Enemigo
class Enemigo:
    def __init__(self, nombre, salud, dano):
        self.nombre = nombre
        self.salud = salud
        self.dano = dano

    def atacar(self):
        return random.randint(5, 15)

    def recibir_dano(self, dano):
        self.salud -= dano
        if self.salud <= 0:
            self.salud = 0
            print(f"Has derrotado a {self.nombre}")
```

```python

import random

def main():
    nombre_jugador = input("Ingrese su nombre: ")
    jugador = Jugador(nombre_jugador)

    # Datos de los enemigos: (nombre, salud, ataque)
    tipos_enemigos = [
        ("Golem", 50, 10),
        ("Dragon", 30, 10),
        ("Ladron", 20, 10),
    ]

    print("Comienza el juego")

    while True:
        # Se crea un enemigo nuevo en cada encuentro
        nombre, salud, ataque = random.choice(tipos_enemigos)
        enemigo_actual = Enemigo(nombre, salud, ataque)
        print(f"Te encontraste a un enemigo: {enemigo_actual.nombre}")

        huyo = False

        # Combate
        while enemigo_actual.salud > 0:
            accion = input("¿Qué deseas hacer? (atacar/huir): ").lower()

            if accion == "atacar":
                dano_jugador = jugador.atacar()
                print(f"Has atacado a {enemigo_actual.nombre} y le causaste {dano_jugador} de daño")
                enemigo_actual.recibir_dano(dano_jugador)

                if enemigo_actual.salud > 0:
                    dano_enemigo = enemigo_actual.atacar()
                    print(f"{enemigo_actual.nombre} te atacó y te hizo {dano_enemigo} de daño")
                    jugador.recibir_dano(dano_enemigo)

                    if jugador.salud <= 0:
                        print("Has perdido la partida")
                        return  # termina el juego

            elif accion == "huir":
                print("Has decidido huir del combate")
                huyo = True
                break

            else:
                print("Acción no válida, escribí 'atacar' o 'huir'")

        # Si ganó el combate, recibe experiencia
        if not huyo:
            print(f"Has derrotado a {enemigo_actual.nombre}")
            jugador.ganar_experiencia(20)

        # Estado después del combate
        print("Estado del jugador:")
        print(f"{jugador.nombre} - Salud: {jugador.salud}, Nivel: {jugador.nivel}, Experiencia: {jugador.experiencia}")

        continuar = input("¿Quieres seguir explorando? (s/n): ").lower()
        if continuar != "s":
            print("Gracias por jugar")
            break

main()
```
